```python
from qoptcraft import Fock
from qoptcraft.optimization import HeraldedStatePrep, GubarevCost
from piquasso_optimization import PiquassoOptimizer

import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from results_table import benchmarks_table
```

# Bell

```python
in_state = Fock(1,1,1,1,0,0)
target_state = Fock(1,0,1,0) + Fock(0,1,0,1)
herald = Fock(1,1)
problem = HeraldedStatePrep(in_state, target_state, herald)
cost_fun = GubarevCost(problem)
opt = PiquassoOptimizer(in_state, target_state, herald)
```

```python
N_RUNS = 1000
results = [opt.minimize() for _ in range(N_RUNS)]
all_results = {"Piquasso": results}
```

```python
PROB_THRESH = 0.073
benchmarks_table(all_results, problem, PROB_THRESH)
```

# Optimizer comparison

```python

from qoptcraft import Fock
from qoptcraft.optimization import (
    HeraldedStatePrep,
    PiecewiseInfidelityCost,
    BFGS,
    LBFGS,
    GradientDescent,
    Adam,
    MomentumGradientDescent,
)

import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from results_table import benchmarks_table
```

```python
riemannian_optimizers = {
    "Riemannian GD": GradientDescent(max_iter=50000, step=0.5),
    "Riemannian Adam": Adam(max_iter=50000, step=0.02),
    "Riemannian NAG": MomentumGradientDescent(max_iter=50000),
    "Riemannian BFGS": BFGS(max_iter=5000, line_search="wolfe"),
    "Riemannian LBFGS": LBFGS(max_iter=5000, line_search="wolfe", memory=10),
}
```

# Bell

```python
in_state = Fock(1,1,1,1,0,0)
target_state = Fock(1,0,1,0) + Fock(0,1,0,1)
herald = Fock(1,1)
problem = HeraldedStatePrep(in_state, target_state, herald)
cost_fun = PiecewiseInfidelityCost(problem)
```

```python
N_RUNS = 1000
all_results = {
    name: [
        optimizer.minimize(cost_fun, cost_fun.manifold.random_point(), verbose=False)
        for _ in range(N_RUNS)
    ]
    for name, optimizer in riemannian_optimizers.items()
}
```

```python
PROB_THRESH = 0.073
benchmarks_table(all_results, problem, PROB_THRESH, time_format=".4f")
```

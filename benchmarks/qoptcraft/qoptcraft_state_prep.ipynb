```python
from qoptcraft import Fock
from qoptcraft.optimization import (
    HeraldedStatePrep,
    GubarevCost,
    BFGS,
    EuclideanOptimizer,
    clements_haar_params,
    algebra_cayley_params,
    algebra_cayley_haar_params,
)

import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from results_table import benchmarks_table
```

```python
riemannian = BFGS(max_iter=5000, line_search="wolfe")
euclidean = EuclideanOptimizer(options={"maxiter": 10000, "gtol": 1e-10})
euclidean_algebra = EuclideanOptimizer(
    options={"maxiter": 10000, "gtol": 1e-10}, parametrize=algebra_cayley_params
)


def run_riemannian(cost_fun):
    """One run optimizing the unitary itself, on the manifold."""
    init_point = cost_fun.manifold.random_point()
    return riemannian.minimize(cost_fun, init_point, verbose=False)


def run_euclidean(cost_fun):
    """One run optimizing the beamsplitter and phase angles of a Clements circuit."""
    init_params = clements_haar_params(cost_fun.manifold.modes)
    return euclidean.minimize(cost_fun, init_params, verbose=False)


def run_euclidean_algebra(cost_fun):
    """One run optimizing the algebra coordinates of a Cayley-parametrized unitary."""
    init_params = algebra_cayley_haar_params(cost_fun.manifold.modes)
    return euclidean_algebra.minimize(cost_fun, init_params, verbose=False)
```

# Bell

```python
in_state = Fock(1,1,1,1,0,0)
target_state = Fock(1,0,1,0) + Fock(0,1,0,1)
herald = Fock(1,1)
problem = HeraldedStatePrep(in_state, target_state, herald)
cost_fun = GubarevCost(problem)
```

```python
N_RUNS = 10000
all_results = {
    "Riemannian BFGS": [run_riemannian(cost_fun) for _ in range(N_RUNS)],
    "Euclidean BFGS": [run_euclidean(cost_fun) for _ in range(N_RUNS)],
    "Euclidean BFGS (algebra)": [run_euclidean_algebra(cost_fun) for _ in range(N_RUNS)],
}
```

```python
PROB_THRESH = 0.073
benchmarks_table(all_results, problem, PROB_THRESH, time_format=".4f")
```

# Invariant optimizer comparison

```python
from qoptcraft import Fock
from qoptcraft.math import haar_random_unitary
from qoptcraft.optimization import (
    ExactStatePrep,
    InvariantExactStatePrep,
    InfidelityCost,
    BFGS,
)

import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from results_table import benchmarks_table
```

```python
optimizer = BFGS(max_iter=1000, tol_grad=1e-9)
```

# Fock |1,1,2,0⟩

```python
in_state = Fock(1,1,2,0)
hidden_unitary = haar_random_unitary(in_state.modes, seed=42)
target_state = in_state.evolution_scattering(hidden_unitary, return_pure_state=True)
problems = {
    "Full U(m)": ExactStatePrep(in_state, target_state),
    "Invariant": InvariantExactStatePrep(in_state, target_state),
}
invariant = problems["Invariant"]
print(f"Blocks {invariant.block_sizes} — free parameters: {in_state.modes**2} -> {invariant.manifold.dim}")
```

```python
N_RUNS = 1000
all_results = {
    name: [
        optimizer.minimize(InfidelityCost(problem), problem.manifold.random_point(), verbose=False)
        for _ in range(N_RUNS)
    ]
    for name, problem in problems.items()
}
```

```python
benchmarks_table(all_results, problems, time_format=".4f", index_name="Search space")
```

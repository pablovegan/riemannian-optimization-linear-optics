```python
from qoptcraft import Fock
from qoptcraft.optimization import (
    HeraldedStatePrep, GubarevCost,
)
from mrmustard_state_prep import MrMustardOptimizer

import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from results_table import benchmarks_table
```

## BELL - 6 modes, 4 photons

```python
in_state = Fock(1,1,1,1,0,0)
target_state = Fock(1,0,1,0) + Fock(0,1,0,1)
herald = Fock(1,1)
problem = HeraldedStatePrep(in_state, target_state, herald)
cost_fun = GubarevCost(problem)
opt = MrMustardOptimizer(in_state, target_state, herald)
```

```python
N_RUNS = 1000
results = [opt.minimize() for _ in range(N_RUNS)]
all_results = {"MrMustard": results}
```

```python
prob_thresh = 0.073
infid_thresh = 1e-7

benchmarks_table(all_results, problem, prob_thresh, infid_thresh)
```

## NOON - 4 modes, 6 photons

```python
in_state = Fock(2,2,2,0)
target_state = Fock(5,0) + Fock(0,5)
herald = Fock(1,0)
problem = HeraldedStatePrep(in_state, target_state, herald)
cost_fun = GubarevCost(problem)

NOON_CUTOFFS = [6, 6, 3, 2]   # exact for herald subspace (n2=1, n3=0); ~5x faster than [7,7,7,7]
opt = MrMustardOptimizer(in_state, target_state, herald, cutoffs=NOON_CUTOFFS)
```

```python
N_RUNS = 1000
results = [opt.minimize() for _ in range(N_RUNS)]
all_results = {"MrMustard": results}
```

```python
PROB_THRESH = 0.056
infid_thresh = 1e-7

benchmarks_table(all_results, problem, prob_thresh, infid_thresh)
```
